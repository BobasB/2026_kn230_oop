## Функції та лямбда вирази
- в будь-якій мові програмування, якщо коди виконується більше одного разу, його краще винести в окрему функцію. Це дозволяє уникнути дублювання коду та полегшує його підтримку.

```python
a = 1 + 1
b = a + 2
print(b)
# це є дублюванням коду, тому краще винести його в окрему функцію
a = 1 + 1
b = a + 2
print(b)
```

```python
# Python не є строго типізованою мовою програмування, 
# але ми можемо прописувати типи аргументів та типи повернення функції. 
# Це є декларація функції, код ще не виконується, а лише описується її сигнатура.
def function_name(arg_1: int, arg_2: str) -> list[str]:
    """
    Функція приймає два аргументи: ціле число та рядок,
    повертає рядок, який є сумою аргументів.
    """
    a = arg_1 + 1
    b = str(a) + arg_2
    print(b) # Цей прінт не спрацьовує якщо ми не викликаємо функцію
    return [b]
```

```python
print(function_name)
function_name(1, "2") # виклик функції з позиційними аргументами
function_name(arg_2="2", arg_1=1) # виклик функції з іменованими аргументами

# оскільки наша функція повертає значення, ми можемо присвоїти його змінній
result = function_name(1, "+_2_=")
print(f"Результат виконання функції: {result}")
```

```python
def resolve_mark_fron_number(mark: int) -> str:
    if mark == 5:
        return "Відмінно"
    elif mark == 4:
        return "Добре"
    elif mark == 3:
        return "Задовільно"
    else:
        return "Незадовільно"

list_of_student_marks = [("Іван", 2), ("Сергій", 3), ("Анастасія", 4), ("Максим", 5)]

for student in list_of_student_marks:
    name, mark = student
    # за допомогою функції ми винесли великий блок коду та оптимізували його
    print(f"Студент {name} отримав оцінку у національній шкалі: {resolve_mark_fron_number(mark)}")
```

- в загальному - функція - це винесення блоку коду в окрему одиницю, яка може бути викликана з різних місць програми.
- для простих функцій існує простий синтаксис - лямбда вирази. Лямбда вирази дозволяють створювати анонімні функції, які можна використовувати в місцях, де потрібна функція, але не потрібно її зберігати в змінній.
- найчастіше лямбда вирази виконуються в одностічкових виразах, але можуть бути і багаторядковими. Лямбда вирази можуть приймати будь-яку кількість аргументів, але можуть містити лише один вираз.
- Приклад - ми маєм нікнейм і потрібно сформувати з нього email адресу. 

```python
users_nicknems = ["Максим", "Анастасія", "Сергій", "Іван"]

def create_email(nickname: str) -> str:
    """Клачсичний запис функції, яка приймає нікнейм та повертає email адресу"""
    return f"{nickname.lower()}@itcollege.com"

# лямбда вирази дозволяють створювати анонімні функції
# def function_name - заміняється на lambda
lf = lambda nickname: f"{nickname.lower()}@itcollege.com"

users_emails_with_lambdas = [lf(nickname) for nickname in users_nicknems]
users_emails_with_function = [create_email(nickname) for nickname in users_nicknems]

print(lf, users_emails_with_lambdas)
print(create_email, users_emails_with_function)
```

```python
lfa = lambda nickname: f"{nickname.lower()}@gmail.com"
lfb = lambda nickname: f"{nickname.lower()}@microsoft.com"

print("Функції є безіменними, і завжди називаються lambda:", lfa, lfb)
```

### Декоратори
- якщо функція не викликана то ми можемо передати її як аргумент в іншу функцію. Це дозволяє створювати декоратори - функції, які приймають інші функції як аргументи та повертають нові функції з додатковою поведінкою.

```python
def decorator_function(another_function):
    """
    Декоратор - функція, яка приймає іншу функцію як аргумент та повертає нову функцію.
    """
    def wrapper_function(*args, **kwargs):
        print("Декоратор виконується перед викликом функції")
        result = another_function(*args, **kwargs)
        print("Декоратор виконується після виклику функції")
        return result
    return wrapper_function
```

```python
@decorator_function
def say_hello(name: str) -> str:
    print(f"Привіт, {name}!")
    return f"Привіт, {name}!"

say_hello("Максим")
```

- рекурсія функцій
- приклад - обчислення чисел фібоначі

```python
def fibonacci(n: int) -> int:
    """Рекурсивна функція для обчислення чисел Фібоначчі"""
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(10))
```
