## Цикли в Python

Цикли в Python дозволяють виконувати блок коду кілька разів, поки виконується певна умова. У Python є два основних типи циклів: `for` і `while`.
- цикл `for` використовується для ітерації по послідовностях (наприклад, списках, кортежах, рядках) або інших ітерованих об'єктах.
- цикл `while` виконує блок коду, поки умова є істинною.

ми купили пачку ММдмс - і хочемо знати скільки раз ми витягнемоя червоний колір
Припустимо що в пачці буде 50 штук

Робимо дану симуляцію двома способами: за допомогою циклу `for` та циклу `while`.

```python
import random
mmds = ["Червоний", "Жовтий", "Зелений", "Синій", "Фіолетовий", "Коричневий"]

#print(list(range(50)))
n_red = 0
for i in range(50):
    m = random.choice(mmds)
    #print(f"Ми витягнули: {m} ММДмс")
    if m == "Червоний":
        print("!!!>> Ура! Ми витягнули червоний колір! <<!!!")
        n_red += 1
print(f"Ми витягнули червоний колір {n_red} разів")
```

Для умовного циклу while ми можемо дуже лугко зайти в нескінченний цикл, тому потрібно обов'язково передбачити умову виходу з нього.
тобто - при вокристанні ітераційної змінної ми повинні памятати про те що вона повинна змінюватися в процесі виконання циклу, інакше ми отримаємо нескінченний цикл.

```python
i = 0
n_red = 0

while i < 50:
    m = random.choice(mmds)
    #print(f"Ми на {i} кроці симуляції витягнули: {m} ММДмс")
    if m == "Червоний":
        print("!!!>> Ура! Ми витягнули червоний колір! <<!!!")
        n_red += 1
    i += 1
print(f"Ми витягнули червоний колір {n_red} разів")
```

у нас є кубик і ми будемо його кидати допоки не випаде 6. Ми хочемо знати скільки разів ми кидали кубик до того моменту поки не випало 6.
якщо є умова, то найкращим варіантом буде використання циклу while, бо ми не знаємо скільки разів нам доведеться кидати кубик, щоб випало 6.

```python
print(random.randint(1, 6))

while random.randint(1, 6) != 6:
    # ітерації повторюються доки умова не стане False, тобто доки не випаде 6
    print("Ми кинули кубик і не випало 6, кидаємо ще раз...")
else: # остання ітерація коли умова стає False
    print("Ура в нас випала 6!!!")
```

переписуємо кидання кубика за допомогою циклу for, але в даному випадку ми не знаємо скільки разів нам доведеться кидати кубик, щоб випало 6, тому цикл for не підходить для даної задачі.
