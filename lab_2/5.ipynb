- є помилки які є прихованими та неочевидними. Найчастіше такі помилки виникають при роботі з користувачами або їх даними. Наприклад, якщо користувач вводить дані у неправильному форматі, або якщо дані зберігаються у неправильному форматі.

````python
## Виловлювання помилок у коді
#a = "1 # SyntaxError: 
# при введенні нечислових значень виникне помилка ValueError
from_user_input = input("Введіть просте ціле число: ")
if not from_user_input.isdigit():
    print("Ви ввели не число!")
print(f"Ви ввели число: {from_user_input}")
````

- для виловлювання помилок про які ми можемо не здогадуватися, у Python є конструкція try-except. Вона дозволяє нам спробувати виконати певний код, і якщо виникає помилка, ми можемо обробити її у блоці except.

````python
import random

for i in range(20):
    try:
        n = random.randint(0, 5)
        print(f"Випадкове число: {n}")
        print(f"Результат ділення {i} на {n} = {i/n}")
    except ZeroDivisionError:
        print("Опрацьовування помилки ZeroDivisionError")
        print(f"Помилка: ділити {i} на {n} - неможливо за математичними законами")

print("Програма завершила роботу")
````

- якщо ми будемо звертатись до даних яких неіснує, то виникне помилка IndexError. Наприклад, якщо ми спробуємо звернутись до елементу списку за індексом який неіснує, то виникне помилка IndexError: list index out of range.

```python
---------------------------------------------------------------------------
IndexError                                Traceback (most recent call last)
Cell In[2], line 5
      1 #import boto3
      3 li = [1, 2]
----> 5 print(li[2]) # IndexError: list index out of range

IndexError: list index out of range
```

````python
#import boto3

li = [1, 2]
#     ^0 ^1

#print(li[2]) # IndexError: list index out of range

for i in range(5):
    try:
        print(f"Ітерація {i+1}. li[{i}] = {li[i]}")
    except IndexError:
        print("Опрацьовування помилки IndexError")
        print(f"Ітерація {i+1}. У списку є лише {len(li)} елементів")

print("Програма завершила роботу")
````
