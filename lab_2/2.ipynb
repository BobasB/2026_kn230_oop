## Вбудовані константи

- дві константи відносяться до логічного типу даних, boolean:
  - `True` - логічне значення "істина"
  - `False` - логічне значення "хибність"

````python
#a1 = true
a2 = True
true = True # сворюємо локальну змінну з ім'ям true, яка містить значення True

#b1 = false
b2 = False
false = False # сворюємо локальну змінну з ім'ям false, яка містить значення False

print("Значення змінної a2:", a2, type(a2))
````

- дані константи використовуються для порівняння, логічних операцій, умовних операторів та інших конструкцій програмування. і все що не є пустим значенням, вважається істинним (True), а пусте значення - хибним (False).
- порожній масив відповідає False тому що у Python ми можемо вибирати елемети масиву допоки він не пустий, а порожній масив не має елементів для вибору, тому він вважається хибним (False).

````python
s1 = "Це не порожнє!"
s2 = ""
if s2:
    print("Змінна порожня, тому вона вважається хибною (False)")

le = list([])
if le:
    print("Список не порожній, тому він вважається істинним (True)")

de = dict({})
if de:
    print("Словник не порожній, тому він вважається істинним (True)")
````

- значення істинності або хибності можна порівнювати з числами, де 0 вважається хибним (False), а будь-яке інше число - істинним (True).

````python
print("Порівняння значень істинності та хибності з числами:", 
      "\nTrue == 1:", True == 1, 
      "\nFalse == 0:", False == 0, 
      "\nTrue == 2:", True == 2, 
      "\nFalse == -1:", False == -1)
    
if 2:
    print("Число 2 вважається істинним (True)")
````

- None - це зарезервоване ключове слово в Python, яке представляє відсутність значення або порожнє значення. Воно використовується для позначення того, що змінна не має жодного значення або що функція не повертає жодного результату.

````python
n1= None

print("Значення змінної n1:", n1, type(n1))
````

````python
return_values = list([print("Це функція print()"), print("ВИВІД")])

print("Повернені значення:", return_values)
````

- повернення з функції значення None відбувається автоматично, якщо функція не містить оператора return або якщо оператор return не повертає жодного значення.
- це сигналізує що функція завершила свою роботу, але не повернула жодного результату.

````python
# приклад функції, є тіло і те що вона повертає
def my_function():
    # тіло функції
    print("Це функція my_function()")
    i = 1 # результат обчислення
    return i

result = my_function()
print("Повернене значення з функції my_function():", result)

def my_function_2():
    # тіло функції
    print("Це функція my_function_2()")
    #raise Exception("Це виключення, яке піднімається у функції my_function_2()")

result = my_function_2()
print("Повернене значення з функції my_function_2():", result)
````

- оскільки порожні масиви відповідають False, то константа None також відповідає False, оскільки вона представляє відсутність значення або порожнє значення.

````python
if None:
    print("Це ніколи не виконається, оскільки None вважається хибним (False)")

# Заперечення пустоти буде істинним (True), 
if not None:
    print("Це виконається, оскільки заперечення None вважається істинним (True)")
````

- ніколи не можна переприсвоювати значення констант True, False та None, оскільки вони є зарезервованими ключовими словами в Python. Неможна створити змінну з іменем True, False або None, оскільки це призведе до помилки синтаксису.

# Вбудовані функції
- детально розглянемо функцію `print()`, яка використовується для виведення інформації на екран. Вона може приймати будь-яку кількість аргументів, які будуть виведені у вигляді рядків.

- функція завжди приймає аргументи, які будуть виведені на екран, і повертає значення None. Це означає, що функція виконує свою роботу (виводить інформацію), але не повертає жодного результату.

````python
print("Тут лише один аргумент")
print() # тут взагалі аргументів немає
print("Це перший аргумент", "Це другий аргумент", "Це третій аргумент")
print()
print()
print()
print("Код зверху найчастіше генурує АІ помічник")
print("\n\n\n") # таки запис роблять розробники, щоб вивести кілька порожніх рядків
print("Нижня межа")
````
