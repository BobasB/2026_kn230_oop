## Вбудовані константи

- дві константи відносяться до логічного типу даних, boolean:
  - `True` - логічне значення "істина"
  - `False` - логічне значення "хибність"

````python
#a1 = true
a2 = True
true = True # сворюємо локальну змінну з ім'ям true, яка містить значення True

#b1 = false
b2 = False
false = False # сворюємо локальну змінну з ім'ям false, яка містить значення False

print("Значення змінної a2:", a2, type(a2))
````

- дані константи використовуються для порівняння, логічних операцій, умовних операторів та інших конструкцій програмування. і все що не є пустим значенням, вважається істинним (True), а пусте значення - хибним (False).
- порожній масив відповідає False тому що у Python ми можемо вибирати елемети масиву допоки він не пустий, а порожній масив не має елементів для вибору, тому він вважається хибним (False).

````python
s1 = "Це не порожнє!"
s2 = ""
if s2:
    print("Змінна порожня, тому вона вважається хибною (False)")

le = list([])
if le:
    print("Список не порожній, тому він вважається істинним (True)")

de = dict({})
if de:
    print("Словник не порожній, тому він вважається істинним (True)")
````

- значення істинності або хибності можна порівнювати з числами, де 0 вважається хибним (False), а будь-яке інше число - істинним (True).

````python
print("Порівняння значень істинності та хибності з числами:", 
      "\nTrue == 1:", True == 1, 
      "\nFalse == 0:", False == 0, 
      "\nTrue == 2:", True == 2, 
      "\nFalse == -1:", False == -1)
    
if 2:
    print("Число 2 вважається істинним (True)")
````

- None - це зарезервоване ключове слово в Python, яке представляє відсутність значення або порожнє значення. Воно використовується для позначення того, що змінна не має жодного значення або що функція не повертає жодного результату.

````python
n1= None

print("Значення змінної n1:", n1, type(n1))
````

````python
return_values = list([print("Це функція print()"), print("ВИВІД")])

print("Повернені значення:", return_values)
````

- це сигналізує що функція завершила свою роботу, але не повернула жодного результату.

````python
# приклад функції, є тіло і те що вона повертає
def my_function():
    # тіло функції
    print("Це функція my_function()")
    i = 1 # результат обчислення
    return i

result = my_function()
print("Повернене значення з функції my_function():", result)

def my_function_2():
    # тіло функції
    print("Це функція my_function_2()")
    #raise Exception("Це виключення, яке піднімається у функції my_function_2()")

result = my_function_2()
print("Повернене значення з функції my_function_2():", result)
````

- оскільки порожні масиви відповідають False, то константа None також відповідає False, оскільки вона представляє відсутність значення або порожнє значення.

````python
if None:
    print("Це ніколи не виконається, оскільки None вважається хибним (False)")

# Заперечення пустоти буде істинним (True), 
if not None:
    print("Це виконається, оскільки заперечення None вважається істинним (True)")
````

# Вбудовані функції
- детально розглянемо функцію `print()`, яка використовується для виведення інформації на екран. Вона може приймати будь-яку кількість аргументів, які будуть виведені у вигляді рядків.

- функція завжди приймає аргументи, які будуть виведені на екран, і повертає значення None. Це означає, що функція виконує свою роботу (виводить інформацію), але не повертає жодного результату.

````python
print("Тут лише один аргумент")
print() # тут взагалі аргументів немає
print("Це перший аргумент", "Це другий аргумент", "Це третій аргумент")
print()
print()
print()
print("Код зверху найчастіше генурує АІ помічник")
print("\n\n\n") # таки запис роблять розробники, щоб вивести кілька порожніх рядків
print("Нижня межа")
````

- форматування аргументів функції `print()`
- аргумент - це будь-яке значення, яке передається функції для обробки. У випадку функції `print()`, аргументи можуть бути рядками, числами, списками, словниками та іншими типами даних.

````python
arg1 = "Це перший аргумент"
print(arg1, "Це другий аргумент", "Це третій аргумент", 1, [1, 2, 3])
````

- ми можемо форматувати аргументи до того як вони будуть виведені на екран, використовуючи різні властивості стрічок та типів даних
- в січках ми рожемо робити конкатенацію (об'єднання) рядків за допомогою оператора `+`
- в print пробіли між аргументами додаються автоматично, тому якщо ми хочемо об'єднати рядки без пробілів, ми можемо використовувати конкатенацію рядків за допомогою оператора `+`.

````python
ak = "Це буде початок"
ab = "та продовження рядка"
print(ak, ab)

a_final = ak + ab
print(a_final) # все передали одним аргументом

b_final = ak + " " + ab
print(b_final)
````

- конкатенація з векикою кількістю аргументів може бути неефективною і навіть важкою

````python
greeting = "Привіт"
name = "Богдан"
sender = "Відправник"

header = greeting + " " + name
main_text = "Запрошую на лекції по об'єктно-орієнтованому програмуванню. Буде цікаво та пізнавально!"
footer = "З повагою, " + sender

final_message = header + "\n" + main_text + "\n" + footer

print(final_message)
````

- ми можемо використати властивості обєкта стрічок і використати метод `join()`, або метод `format`

````python
l_words = ["Це", "буде", "рядок", str(1),  "з", "списку"]
final_string = " | ".join(l_words)
print(final_string, type(final_string))

gret = " ".join([greeting, name])
print(gret, type(gret))
````

- через format - ми бачимо пропущені аргументи, які будуть замінені на значення змінних, що передаються у форматі. Це дозволяє створювати більш гнучкі та динамічні рядки, де значення можуть змінюватися в залежності від контексту.
- є можливіть робити позиційне форматування, де ми можемо вказати індекси аргументів у форматі, щоб визначити порядок їх вставки у рядок. Це дозволяє нам контролювати, які значення будуть вставлені у рядок та в якому порядку.

````python
fm = "{} {}! {}е раді бачити Вас. {} чи готові вчитися?".format(greeting, name, name, name)
print(fm, type(fm))

fpm = "{0} {1}! {1}е раді бачити Вас. {1} чи готові вчитися?".format(greeting, name)
print(fpm, type(fpm))
````

- f-strings (форматовані рядки) - це ще один спосіб форматування рядків у Python. Вони дозволяють вставляти значення змінних безпосередньо у рядок за допомогою фігурних дужок `{}`. Це робить код більш читабельним та зручним для розуміння.
- у таких стрічках використовуємо іменні змінні

````python
multyline_str = f"""
Перший рядок привітання
{greeting} {name}!
{name} раді бачити Вас. {name} чи готові вчитися?

Основне частина повідомлення

Завершення
З повагою,
{sender}
"""

print(multyline_str)
````

- в строго типізованих мовах програмування, таких як C++ або Java, типи даних повинні бути визначені заздалегідь - тобто скільки і який тип даних буде використовуватися у функції. У Python, який є динамічно типізованою мовою, типи даних визначаються під час виконання програми, що дозволяє більш гнучко працювати з різними типами даних.
- для повного контролю над кодом - ми форматуємо рядки за допомогою f-strings, що дозволяє нам вставляти значення змінних безпосередньо у рядок та контролювати форматування виводу.
