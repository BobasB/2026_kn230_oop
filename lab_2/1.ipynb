## Типи даних
- прості типи даних і починаємо з числових типів даних:
  - int (цілі числа)
  - float (дробові числа)

```python
a = 1
b = 1.1
print("Значення змінної a:", a, "Тип:", type(a))
print("Значення змінної b:", b, "Тип:", type(b))

c = a + b
print("Значення змінної c:", c, "Тип:", type(c))
```

- неявна та явна трансформація типів даних:
  - неявна трансформація (implicit type conversion) - автоматична зміна типу даних під час виконання операцій
  - явна трансформація (explicit type conversion) - зміна типу даних за допомогою функцій, таких як int(), float(), str() тощо

```python
d = int(b) # явне перетворення типу даних
e = b + d # неявне перетворення типу даних
print("Значення:", d, b, e, "Типu:", type(d), type(b), type(e))
f = int(b + d) # явне перетворення типу даних
print("Значення:", f, "Тип:", type(f))
```

```python
def my_function(i: int):
    # якщо робити оперецію з int і float, то результат буде float, тому потрібно явно перетворювати результат на int
    return i
```

## Обєкти і Python
- Все в Python є об'єктом, тому всі типи даних є об'єктами. Кожен об'єкт має свій тип, який можна визначити за допомогою функції type().
- оскільки ми знаємо що Python є динамічно типізованою мовою, і при створенні змінних вони також є об'єктами то ми можемо звертатись до них як до змінних і як до об'єктів, тобто ми можемо викликати методи об'єктів, які є змінними. Наприклад, ми можемо викликати методи об'єкта int, який є змінною a.

```python
a1 = int(1)
a2 = 2.5
a3 = a1 + a2

# та ж сама операція але через обєкти
a4 = a2.__add__(a1)
a5 = a1.__add__(a2)
print("Значення:", a3, a4, a5, "Типи:", type(a3), type(a4), type(a5))

#print(NotImplemented + 1)
print(type(NotImplemented))
```

## Змінні стічкового типу str
- все що приєвоюється в лапках " або ' є рядком, тобто змінною стічкового типу str.
- багатострічковий рядок можна створити за допомогою потрійних лапок """ або '''.

```python
s1 = "Ми вчимось Python"
print("Значення змінної s1:", s1, "Тип:", type(s1))
s2 = 2
s3 = '2'

# s2 + s3 # це дасть помилку, бо не можна додавати int і str
# для таких виапдків потрібно явно перетворювати типи даних
i4 = s2 + int(s3)
print("Значення:", i4, "Тип:", type(i4))
```

## Операції на різними типами даних
- найчастіше це арифметивні операції які змінюють властивості залежно відтипу даних з яким вони виконуються. Наприклад, при додаванні int і float результат буде float, а при додаванні str і int буде помилка.

```python
i1 = 1
i2 = 2
i3 = i1 + i2
i4 = i1 * i2
print("Значення змінної i3:", i3, "Тип:", type(i3), type(i4))
# автоматична конвертація типів даних при виконанні операцій через фізичні комірки памяті
i5 = i4 + 5.1
print("Значення змінної i5:", i5, "Тип:", type(i5))
```

- конкатенація (concatenation) - операція об'єднання рядків, яка виконується за допомогою оператора +.
- її можна виконувати з числами, але результат буде рядок, тобто str.
- оскільки Python не знає з яким кінцевим типом даних ми хочемо працювати, то ми маємо явно вказати тип даних, до якого ми хочемо привести змінну, тобто виконати явне перетворення типу даних.

```python
s1 = "Ми вчимо "
s2 = "Python"
s3 = s1 + s2
print("Значення змінної s3:", s3, "Тип:", type(s3))

s4 = "P" + "y" + "t" + "h" + "o" + "n"
print("Значення змінної s4:", s4, "Тип:", type(s4))

i1 = 1
s1 = "2"
#s3 = i1 + s1 # це дасть помилку, бо не
print(i1 + int(s1), " АБО ", str(i1) + s1) # явне перетворення типу даних
```

```python
i1 = 5
i2 = "2"

i = i1 - int(i2)
print("Значення:", i, "Тип:", type(i))
#j = str(i1) - i2 # Це помилка так неможна
```

в інших мовах програмування опереції над різними типами даних можуть бути заборонені, або ми їх можемо перевизначити зробивши перевантаження операторів, але в Python це не потрібно, бо Python сам знає як виконати операцію над різними типами даних.

індексація (indexing) - операція доступу до елементів послідовності за їх індексом. У Python індексація починається з нуля, тобто перший елемент має індекс 0, другий - 1 і так далі. Для доступу до елементів рядка або списку використовується квадратні дужки [].

```python
s1 = " Python"
print(s1, "або поелементно:", s1[0], s1[1], s1[2], s1[3], s1[4], s1[5], s1[6])
```

- зворотні індекси (negative indexing) - операція доступу до елементів послідовності з кінця за їх індексом. У Python зворотні індекси починаються з -1, тобто останній елемент має індекс -1, передостанній - -2 і так далі. Для доступу до елементів рядка або списку використовується квадратні дужки [].
- (ЗАДАЧА СПІБЕСІДИ) в інших мовах програмування для доступу до останнього елемента послідовності потрібно знати її довжину, але в Python це не потрібно, бо ми можемо використовувати зворотні індекси.

```python
s1 = " Python END"
print(s1[6], s1[len(s1)-1], s1[-1])
```

- slicing (нарізка) - операція доступу до підпослідовності елементів послідовності за їх індексами. У Python нарізка виконується за допомогою квадратних дужок [] та двокрапки :. Перший індекс завжди входить в підпослідовність, а другий індекс завжди оминається.
- є неперервні слайси або можна використовувати крок (step) для доступу до елементів послідовності через певну кількість елементів. Крок задається третім індексом після двокрапки.

```python
s1 = " Python END Але стрічка трохи завелика"
print(s1[1:7], s1[-3:])

print("Хитро ->", s1[0:-1:7])
```
