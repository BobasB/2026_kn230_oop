## Контестний менеджер with у Python
- Контестний менеджер (context manager) в Python дозволяє керувати ресурсами, такими як файли, мережеві з'єднання або блокування, забезпечуючи автоматичне відкриття та закриття ресурсів. Це досягається за допомогою ключових слів `with` та методів `__enter__` і `__exit__`.
- пробуємо приклад з читання-запису файлу:

````python
f = open("test.txt", "w") # видкриваємо файл для запису
f.write("Hello, world!\n") # записуємо рядок у файл

print("Файл відкрито для запису")
print("Наступна команда впаде бо файл вже відритий іншим процесом")

fd = open("test.txt", "w") # відкриваємо файл для читання
f.write("Це новий рядок від F") # записуємо рядок у файл
fd.write("Це новий рядок від FD") # записуємо рядок у файл
f.close() # закриваємо файл
fd.close() # закриваємо файл
````

- для правильної роботи з файлами, краще використовувати контекстний менеджер `with`, який автоматично закриває файл після завершення роботи з ним. Ось приклад:
- при повторному доступі до файлу він вже буде закритий, і ми будемо мати помилку
```python
ValueError: I/O operation on closed file.
```

````python
with open("test.txt", "w") as fw_1: # відкриваємо файл для запису
    fw_1.write("Hello, world!\n") # записуємо рядок у файл
    print("Файл відкрито для запису")
    print("Наступна команда впаде бо файл вже відритий іншим процесом")

print("Файл автоматично закрито після виходу з блоку with")

with open("test.txt", "w") as fw_2: # відкриваємо файл для запису
    fw_2.write("Hello, world!\n") # записуємо рядок у файл
    print("Файл відкрито для запису")
    print("Наступна команда впаде бо файл вже відритий іншим процесом")

# з цього моменту ми не можемо доступитись до першого разу як ми відрили файл
#fw_1.write("Це новий рядок від F") # записуємо рядок у файл
````

Приклад
багаторазове відриття файлу, коли ми будемо записувати у файл рядки таблиці

````python

with open("table.md", "w") as f: # відкриваємо файл для запису
    # Перший раз відриваємо файл і записуємо заголовок таблиці
    f.write("| Імя | Бал | Буква |\n") # записуємо рядок у файл
    f.write("|------|-------|------|\n") # записуємо рядок у файл
# Далі ми можемо виконувати операції по створенню рядків таблиці
table_data = [("Іван", 5), ("Марія", 4), ("Петро", 3), ("Олена", 5), ("Сергій", 2)]
mark = ["E", "F", "D", "C", "B", "A"]
formatted_row = ""
for i in table_data:
    formatted_row += f"| {i[0]} | {i[1]} | {mark[i[1]]} |\n"

print("Вигляд таблиці:\n", formatted_row)
with open("table.md", "a") as f: # відкриваємо файл для додавання
    f.write(formatted_row) # записуємо рядки таблиці у файл
````
