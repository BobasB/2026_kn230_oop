## Розгалуження (умовні оператори)
- класичне предствалення це - if - else оператори, які дозволяють виконувати різні блоки коду в залежності від певних умов.

```python
condition = bool(1) 

if condition:
    print("Умова істинна")
    print("Виконується блок коду, якщо умова істинна")
    print("Кінець блоку коду, якщо умова істинна")
else:
    print("Умова хибна")
    print("Виконується блок коду, якщо умова хибна")
    print("Кінець блоку коду, якщо умова хибна")

print("Кінець програми")
```

```python
if not condition:
    print("Умова хибна")

print("Умова при фолсе")
```

- one-liners - однострічкові записи умовних операторів, які дозволяють виконувати різні блоки коду в залежності від певних умов, все в одній лінії коду.
- це робиться найчастіше для присвоєння значень змінним, але може бути використано і для виконання функцій.

```python
A = random.choice([True, False])

if A:
    print("Умова істинна")
else:
    print("Умова хибна")

ol = "Умова істинна" if A else "Умова хибна"
print(ol)
```

- йдучи далі і складніше, ми можемов виконати однострічковий цикл з умовним оператором, який дозволяє виконувати різні блоки коду в залежності від певних умов, все в одній лінії коду.
- list comprehension - це спосіб створення списків за допомогою однострічкових циклів з умовними операторами, який дозволяє виконувати різні блоки коду в залежності від певних умов, все в одній лінії коду.

```python
lf = [i for i in range(10) if i % 2 == 0]
print(lf)
colors = ["red", "green", "blue", "yellow", "green", "purple", "green", "orange"]
# фільтр елементів списку за допомогою list comprehension
lf_2 = [color for color in colors if color != "green"]
print(lf_2)
```
