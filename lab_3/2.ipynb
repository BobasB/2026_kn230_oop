## ООП видиміть змінних та атрибутів
- є змінні які видимі та достіпні з обєкта і є такі що доступні з класу.

```python
class Student:
    temp_var = [] # атрибут класу, доступний з класу та обєкта, він не унікальний для обєкта
    var_counter = 0 # атрибут класу, доступний з класу та обєкта, він не унікальний для обєкта
    def __init__(self, name, age):
        self.name = name # атрибути є унікальними для обєкта
        self.age = age
        Student.var_counter += 1 # використовуємо як лічильний
    
    def __del__(self):
        Student.var_counter -= 1 # використовуємо як лічильний
        del self


so1 = Student("Сергій", 20)
so1.temp_var.append("Я унікальний для обєкта so1") # змінна temp_var доступна з обєкта so1
so2 = Student("Олександр", 21)
so2.temp_var[0] = so2.temp_var[0] + " Я унікальний для обєкта so2"
so3 = Student("Іван", 22)

print("Кожен атрибут буде унікальним для свого обєкта:", so1.name, so2.name)
print("Але ми доступаємось з іншого обєкту тогож класу:", so2.temp_var)
del so1
print("Конструктор класу був викликаний", Student.var_counter, "разів")
```

```python
class Student2:
    ALL_STUDENTS = [] # атрибут класу, доступний з класу та обєкта, він не унікальний для обєкта
    def __init__(self, name, age):
        self.name = name # атрибути є унікальними для обєкта
        self.age = age
        Student2.ALL_STUDENTS.append(self) # використовуємо як лічильний

so1 = Student2("Сергій", 20)
so2 = Student2("Олександр", 21)

print("Ім'я першого студента:", Student2.ALL_STUDENTS[0].name)
print("З другого обєкта до першого:", so2.ALL_STUDENTS[0].name) # доступ до атрибуту класу через обєкт
```
