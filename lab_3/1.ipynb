## Обєктно-орієнтоване програмування (ООП)
- до цього ми розгядали вбудовані класи та об'єкти, які надає мова програмування Python. Тепер ми розглянемо, як створювати власні класи та об'єкти.
- В ООП все починається з Класів. Клас - це шаблон (blueprint) для створення об'єктів, який визначає їхні атрибути та методи.
- оскількаи клас - це шаблон, ми можемо самі спроетувати або задизайнувати його, тобто створити власний клас. Для цього використовується ключове слово `class`, після якого йде назва класу (імена класів зазвичай починаються з великої літери).
- як тільки в нас є клас, ми можемо створювати об'єкти цього класу. Уявляйте собі як форму для створення деталей, де клас - це форма, а об'єкти - це конкретні деталі, створені за цією формою.
- Об'єкт - це конкретний екземпляр класу, який має свої власні значення атрибутів та може виконувати методи, визначені в класі.

```python
i = int(1)
print(type(i), i.__class__)
```

```python
class PassEmpty:
    pass

my_object = PassEmpty()
print(type(my_object), my_object.__class__)
```

- в кожному класі є спеціальний метод `__init__`, який викликається автоматично при створенні об'єкта. Він використовується для ініціалізації атрибутів об'єкта. Ще його називають конструктором класу. Він може приймати параметри, які передаються при створенні об'єкта, і використовує їх для встановлення початкових значень атрибутів.

```python
class MyClassWithConstructor:
    def __init__(self):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = "Богдан" # статичні атрибути (константи)
        self.age = 38

for i in range(3):
    my_object = MyClassWithConstructor()
    print(f"Об'єкт {i+1}, його атрибути: name = {my_object.name}, age = {my_object.age}")
```

- за допомогою конструктора - ми можемо параметризувати об'єкти, тобто створювати об'єкти з різними значеннями атрибутів. Це дозволяє нам створювати об'єкти з різними станами та поведінкою, що робить наш код більш гнучким та адаптивним до різних ситуацій.
- для цього в конструкторі ми можемо визначати параметри, які будуть передаватися при створенні об'єкта. Ці параметри можна використовувати для встановлення значень атрибутів об'єкта.

```python
class MyClassWithAttributes:
    def __init__(self, name: str, age: int):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name # динамічні атрибути (параметризовані)
        self.age = age

list_of_names = [("Богдан", 38), ("Сергій", 19), ("Артур", 22)]
list_of_objects = []
for name, age in list_of_names:
    # створюємо об'єкт класу MyClassWithAttributes передаючи всередину параметрами name та age
    c = MyClassWithAttributes(name, age)
    list_of_objects.append(c)
    print(f"Об'єкт з параметрами: name = {c.name}, age = {c.age}")

print("\nВсі об'єкти з параметрами:", type(list_of_objects), type(list_of_objects[0]))
```

- оскільки ми маємо атрибути, ми можемо їх порівнювати.
- для прикладу, визначимо хто є старшим згідно атрибуту `age`. Для цього ми можемо визначити методи порівняння, які будуть порівнювати об'єкти за певним критерієм. У нашому випадку, ми будемо порівнювати об'єкти за атрибутом `age`.
- при взаємодії обєктів, варто разглянути вказівник на об'єкт, тобто посилання на об'єкт яким є змінна self. Це дозволяє нам порівнювати об'єкти та визначати їхні відносини один до одного.

```python
class MyClassWithComparison:
    def __init__(self, name: str, age: int):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name # динамічні атрибути (параметризовані)
        self.age = age
    
    def compare_age(self, other):
        """Порівнює об'єкти за атрибутом age."""
        if self.age > other.age:
            return f"{self.name} старший за {other.name} на {self.age - other.age} років"
        elif self.age < other.age:
            return f"{self.name} молодший за {other.name} на {other.age - self.age} років"
        else:
            return f"{self.name} та {other.name} одного віку"

list_of_objects = []
for name, age in list_of_names:
    c = MyClassWithComparison(name, age)
    list_of_objects.append(c)

me = list_of_objects[0]  # Богдан
for other in list_of_objects[1:]:
    print(me.compare_age(other))
```

- маючи клас, ми інадалі може робити динамічні атрибути, але так робити не бажано
- всі атрибути мають створюватись та ініціалізуватись під час створення об'єкта, тобто в конструкторі класу. 
- значення атрибутів об'єкта можна змінювати після його створення, але це не завжди бажано, оскільки це може призвести до непередбачуваних результатів та ускладнити відстеження стану об'єкта.
- схожими до атрибутів є поля які називаються властивостями (properties). Основна характеристика - вони доступні лише для читання, тобто їх не можна змінювати після створення об'єкта. Властивості дозволяють нам контролювати доступ до даних об'єкта та забезпечувати їхню цілісність.

```python
class Student:
    def __init__(self, name: str, age: int, grade: float):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name
        self.age = age
        self.grade = grade

    @property
    def get_info(self):
        """Повертає інформацію про студента."""
        return f"Студент: {self.name}, вік: {self.age}, оцінка: {self.grade}"

sc1 = Student("Богдан", 38, 4.5) #аргументи можуть бути позиційними або іменними
sc1.age = 39 #можна змінювати атрибути об'єкта після його створення, але це не завжди бажано
for sc in [sc1]:
    print(sc.get_info)
    #sc.get_info = "Нова інформація" # спроба змінити значення властивості, яка не має сеттера, викличе помилку
```

- деколи обєкти використовуються для приховування даних, тому існують публічні та приватні атрибути. Публічні атрибути доступні для читання та запису ззовні класу, тоді як приватні атрибути доступні лише всередині класу. Це дозволяє нам контролювати доступ до даних об'єкта та забезпечувати їхню цілісність.
- в Пайтон приватні атрибути не завжди є прихованими і до них можна доступитись і змінювати їх значення, але це не бажано робити, оскільки це порушує принцип інкапсуляції та може призвести до непередбачуваних результатів.
- public, private, protected - це модифікатори доступу, які визначають рівень доступу до атрибутів та методів класу. Public означає, що атрибут або метод доступний з будь-якого місця в коді, private означає, що атрибут або метод доступний лише всередині класу, а protected означає, що атрибут або метод доступний лише всередині класу та його підкласів.

```python
class StudentWithPrivateAttributes:
    def __init__(self, name: str, age: int, grade: float):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name  # публічний атрибут
        self._age = age    # протектед атрибут
        self.__grade = grade  # приватний атрибут
        self.set_mark(grade)  # приватний атрибут, який встановлюється через метод
    
    @property
    def get_info(self):
        """Повертає інформацію про студента."""
        return f"Студент: {self.name}, вік: {self._age}, оцінка: {self.__mark}"

    def set_mark(self, mark: float):
        """Встановлює оцінку студента."""
        if 0 <= mark <= 5:
            print(f"Оцінка {mark} встановлена для студента {self.name}.")
            self.__mark = mark
        else:
            print(f"Помилка: оцінка {mark} недопустима для студента {self.name}.")
            self.__mark = None
    
sc1 = StudentWithPrivateAttributes("Богдан", 38, 3.5)
print(sc1.get_info)
sc1.name = "Сергій"  # можна змінювати публічний атрибут
sc1._age = 39  # можна змінювати протектед (за конвенціями такого бажано не робити)
sc1.__grade = 6.0  # спроба змінити
sc1.__mark = 6.0  # спроба змінити приватний атрибут, але це не змінить його значення

print(sc1.get_info)
```

- важливим в ООП є те що обєкт не лише вмістилище інформації або даних, а й має поведінку, тобто методи. Методи - це функції, які визначені всередині класу та можуть виконувати певні дії з об'єктом або його атрибутами. Методи дозволяють нам визначати поведінку об'єкта та взаємодіяти з ним.
- робимо міні гру, де кожен гравець може кидати кубик, і потім порівнюємо число на кубику, виграв той гравець у якого число на кубику більше. Для цього ми можемо визначити методи для кидання кубика та порівняння результатів кидка.

```python
class PlayerCube:
    """Клас для представлення гравця з кубиком."""

    def __init__(self, name: str):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name
        self._score = 0  # протектед атрибут для зберігання результату кидка кубика
        self._wins = 0  # протектед атрибут для зберігання кількості перемог

    def roll_dice(self):
        """Метод для кидання кубика та оновлення результату."""
        import random
        self._score = random.randint(1, 6)
        return self._score

    @property
    def score(self):
        """Властивість для отримання результату кидка кубика."""
        return self._score

    @property
    def wins(self):
        """Властивість для отримання кількості перемог."""
        return self._wins

player1_name = input("Введіть ім'я першого гравця: ")
player2_name = input("Введіть ім'я другого гравця: ")

pl1 = PlayerCube(player1_name)
pl2 = PlayerCube(player2_name)

print(f"Перевіряємо що гравці не махлюють: {pl1.name} кидок кубика = {pl1.score}, {pl2.name} кидок кубика = {pl2.score}")
for i in range(5):
    print(f"\nРаунд {i+1}:")
    pl1.roll_dice()
    pl2.roll_dice()
    print(f"{pl1.name} кидок кубика = {pl1.score}, {pl2.name} кидок кубика = {pl2.score}")

    if pl1.score > pl2.score:
        print(f"Виграв гравець: {pl1.name}")
        pl1._wins += 1
    elif pl1.score < pl2.score:
        print(f"Виграв гравець: {pl2.name}")
        pl2._wins += 1
    else:
        print("Нічия!")

print(f"\nПідсумки гри:")
print(f"{pl1.name} переміг {pl1.wins} разів.")
print(f"{pl2.name} переміг {pl2.wins} разів.")
```
