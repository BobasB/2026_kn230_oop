## Обєктно-орієнтоване програмування (ООП)
- до цього ми розгядали вбудовані класи та об'єкти, які надає мова програмування Python. Тепер ми розглянемо, як створювати власні класи та об'єкти.
- В ООП все починається з Класів. Клас - це шаблон (blueprint) для створення об'єктів, який визначає їхні атрибути та методи.
- оскількаи клас - це шаблон, ми можемо самі спроетувати або задизайнувати його, тобто створити власний клас. Для цього використовується ключове слово `class`, після якого йде назва класу (імена класів зазвичай починаються з великої літери).
- як тільки в нас є клас, ми можемо створювати об'єкти цього класу. Уявляйте собі як форму для створення деталей, де клас - це форма, а об'єкти - це конкретні деталі, створені за цією формою.
- Об'єкт - це конкретний екземпляр класу, який має свої власні значення атрибутів та може виконувати методи, визначені в класі.

```python
i = int(1)
print(type(i), i.__class__)
```

```python
class PassEmpty:
    pass

my_object = PassEmpty()
print(type(my_object), my_object.__class__)
```

- в кожному класі є спеціальний метод `__init__`, який викликається автоматично при створенні об'єкта. Він використовується для ініціалізації атрибутів об'єкта. Ще його називають конструктором класу. Він може приймати параметри, які передаються при створенні об'єкта, і використовує їх для встановлення початкових значень атрибутів.

```python
class MyClassWithConstructor:
    def __init__(self):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = "Богдан" # статичні атрибути (константи)
        self.age = 38

for i in range(3):
    my_object = MyClassWithConstructor()
    print(f"Об'єкт {i+1}, його атрибути: name = {my_object.name}, age = {my_object.age}")
```

- за допомогою конструктора - ми можемо параметризувати об'єкти, тобто створювати об'єкти з різними значеннями атрибутів. Це дозволяє нам створювати об'єкти з різними станами та поведінкою, що робить наш код більш гнучким та адаптивним до різних ситуацій.
- для цього в конструкторі ми можемо визначати параметри, які будуть передаватися при створенні об'єкта. Ці параметри можна використовувати для встановлення значень атрибутів об'єкта.

```python
class MyClassWithAttributes:
    def __init__(self, name: str, age: int):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name # динамічні атрибути (параметризовані)
        self.age = age

list_of_names = [("Богдан", 38), ("Сергій", 19), ("Артур", 22)]
list_of_objects = []
for name, age in list_of_names:
    # створюємо об'єкт класу MyClassWithAttributes передаючи всередину параметрами name та age
    c = MyClassWithAttributes(name, age)
    list_of_objects.append(c)
    print(f"Об'єкт з параметрами: name = {c.name}, age = {c.age}")

print("\nВсі об'єкти з параметрами:", type(list_of_objects), type(list_of_objects[0]))
```

- оскільки ми маємо атрибути, ми можемо їх порівнювати.
- для прикладу, визначимо хто є старшим згідно атрибуту `age`. Для цього ми можемо визначити методи порівняння, які будуть порівнювати об'єкти за певним критерієм. У нашому випадку, ми будемо порівнювати об'єкти за атрибутом `age`.
- при взаємодії обєктів, варто разглянути вказівник на об'єкт, тобто посилання на об'єкт яким є змінна self. Це дозволяє нам порівнювати об'єкти та визначати їхні відносини один до одного.

```python
class MyClassWithComparison:
    def __init__(self, name: str, age: int):
        """Це Конструктор класу, який викликається автоматично при створенні об'єкта."""
        self.name = name # динамічні атрибути (параметризовані)
        self.age = age
    
    def compare_age(self, other):
        """Порівнює об'єкти за атрибутом age."""
        if self.age > other.age:
            return f"{self.name} старший за {other.name} на {self.age - other.age} років"
        elif self.age < other.age:
            return f"{self.name} молодший за {other.name} на {other.age - self.age} років"
        else:
            return f"{self.name} та {other.name} одного віку"

list_of_objects = []
for name, age in list_of_names:
    c = MyClassWithComparison(name, age)
    list_of_objects.append(c)

me = list_of_objects[0]  # Богдан
for other in list_of_objects[1:]:
    print(me.compare_age(other))
```
